```python
datetime_cols = ['trip_creation_time','od_start_time','od_end_time','cutoff_timestamp']

for col in datetime_cols:
    df[col] = pd.to_datetime(df[col],format='mixed')
```

- Now that all the columns are fixed we can go ahead with the EDA

```python
# Extracting date features from the Trip Creation date column

df['trip_year'] = df['trip_creation_time'].dt.year
df['trip_month'] = df['trip_creation_time'].dt.month
df['trip_day'] = df['trip_creation_time'].dt.day
df['trip_hour'] = df['trip_creation_time'].dt.hour
df['trip_weekday'] = df['trip_creation_time'].dt.day_name()
```

```python
# Extracting features for Source and destination
source_split = df['source_name'].str.rsplit('_', n=2, expand=True)

df['source_city'] = source_split[0]
df['source_place'] = source_split[1]
df['source_state'] = source_split[2]
```

```python
destination_split = df['destination_name'].str.rsplit('_', n=2, expand=True)

df['destination_city'] = destination_split[0]
df['destination_place'] = destination_split[1]
df['destination_state'] = destination_split[2]
```

- Calculating the time taken between od_start_time and od_end_time and keep it as a feature

```python
df['od_time_diff'] = (df['od_end_time'] - df['od_start_time']).dt.total_seconds() / 60
```

```python
# Dropping both columns as they are not required

df.drop(columns=['od_start_time', 'od_end_time'],inplace=True)
```

```python
# Aggregating the data using trip_uuid,source_center,destination_center

agg_df = df.groupby(['trip_uuid', 'source_center', 'destination_center']).agg({
    'actual_distance_to_destination': 'sum','actual_time': 'sum','osrm_time': 'sum','osrm_distance': 'sum','segment_actual_time': 'sum','segment_osrm_time': 'sum',
    'segment_osrm_distance': 'sum','start_scan_to_end_scan': 'first','od_time_diff': 'sum','route_type': 'first','source_state': 'first',
    'destination_state': 'first'}).reset_index()
```

```python
# Aggregating with trip_uuid only

trip_df = df.groupby('trip_uuid').agg({
    'actual_distance_to_destination': 'sum','actual_time': 'sum','osrm_time': 'sum','osrm_distance': 'sum','segment_actual_time': 'sum','segment_osrm_time': 'sum',
    'segment_osrm_distance': 'sum','start_scan_to_end_scan': 'first','od_time_diff': 'sum','route_type': 'first','source_state': 'first',
    'destination_state': 'first'}).reset_index()
```

# Hypothesis Testing

#A. od_time_diff vs start_scan_to_end_scan

Hypothesis

H0: Mean difference between both times is zero.

H1: Mean difference between both times is significant.

```python
from scipy.stats import ttest_rel
```

```python
stat, p = ttest_rel(
    trip_df['od_time_diff'],
    trip_df['start_scan_to_end_scan']
)

if p > 0.05:
    print('Accept null hypothesis.')
else:
    print('Reject the null hypothesis.')
```

- As per the test we can see that there is significant difference in the time when the package is scanned vs trip start time

# B. Actual Time vs OSRM Time

H0: Actual delivery time equals OSRM estimated time.

H1: Actual delivery time significantly differs from OSRM estimated time.

```python
stat, p = ttest_rel(trip_df['actual_time'],trip_df['osrm_time'])
if p > 0.05:
    print('Accept null hypothesis.')
else:
    print('Reject the null hypothesis.')
```

- Actual delivery time is significantly different from predicted routing engine time.

# C. Actual Time vs Segment Actual Time


H0: No significant difference exists between Actual time and Segmented Time

H1: A significant difference exists between the two.

```python
stat, p = ttest_rel(trip_df['actual_time'],trip_df['segment_actual_time'])

if p > 0.05:
    print('Accept null hypothesis.')
else:
    print('Reject the null hypothesis.')
```

This indicates minor inconsistencies between segment-level operational records and overall trip-level records.

# D. OSRM Distance vs Segment OSRM Distance


H0: No significant difference exists between the ORSM and Segmented ORSM distance

H1:There is a significant difference between the two metrics

```python
stat, p = ttest_rel(trip_df['osrm_distance'],trip_df['segment_osrm_distance'])

if p > 0.05:
    print('Accept null hypothesis.')
else:
    print('Reject the null hypothesis.')


```

The aggregated segment distances differ significantly from total OSRM distance estimates.

# E. OSRM Time vs Segment OSRM Time


H0: No significant difference exists

H1: There is significant difference

```python
stat, p = ttest_rel(trip_df['osrm_time'],trip_df['segment_osrm_time'])

if p > 0.05:
    print('Accept null hypothesis.')
else:
    print('Reject the null hypothesis.')
```

Segment-level OSRM predicted time differs significantly from overall OSRM predicted time.

# Checking for Outliers

```python
num_cols = ['actual_time','osrm_time','actual_distance_to_destination','osrm_distance','segment_actual_time','segment_osrm_time','segment_osrm_distance','od_time_diff','start_scan_to_end_scan']
```

```python
for col in num_cols:

  plt.figure(figsize=(8,4))
  sns.boxplot(x=trip_df[col])
  plt.title(f'Boxplot of {col}')
  plt.show()
```

- Outliers were observed across all of the numerical variables, which is expected in logistics operations due to route variability, operational delays, and long-distance shipments.

# IQR Method

```python
for col in num_cols:

    Q1 = trip_df[col].quantile(0.25)
    Q3 = trip_df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    trip_df = trip_df[
        (trip_df[col] >= lower_limit) &
        (trip_df[col] <= upper_limit)
    ]
```
